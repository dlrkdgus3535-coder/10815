```python
from datetime import datetime, date, time
print('<디데이(D-day)프로그램>')
print()
year = int(input('연도를 입력하세요 :'))
month = int(input('월을 입력하세요 :'))
day = int(input('일을 입력하세요: '))
dt4 = date(year, month, day)
today = date.today( )               # today = 오늘 날짜 저장
dday = (dt4- today).days            # ddat  = 두 날짜 사이의 일수 계산하여 저장
print(f'{year}년 {month}월 {day}일은 {dday}일 남았습니다')
```

```python
a = int(input('빨간색 갯수 :'))
b = int(input('노란색 갯수 :'))
c = int(input('파란색 갯수 :'))
d = int(input('초록색 갯수 :'))
print('=====회전 초밥 영수증=====')
print('  접시  단가  개수  합계')
print(f' 빨간색 1,000  {a}   {1000*a}')
print(f' 노란색 2,000  {b}   {2000*b}')
print(f' 파란색 3,000  {c}   {3000*c}')
print(f' 초록색 4,000  {d}   {4000*d}')
print(f' 총가격          {1000*a+2000*b+3000*c+4000*d}')
print('-----------------------')
```
